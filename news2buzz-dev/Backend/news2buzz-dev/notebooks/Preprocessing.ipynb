```python
import pandas as pd
import numpy as np
import re

def is_truly_blank(series):
    return series.astype(str).apply(lambda x: bool(re.fullmatch(r'\s*', x)))

# Step 1: Count NaN
null_counts = df.isna().sum()

# Step 2: Count blank and invisible entries
blank_counts = df.apply(lambda col: is_truly_blank(col).sum())

# Step 3: Total missing values
total_missing = null_counts + blank_counts

# Step 4: Filter and display only columns with missing
missing_report = total_missing[total_missing > 0]

print("🕵️ Refined Missing/Invisible/Blank Count per Column:\n")
print(missing_report.sort_values(ascending=False))
```

```python
# 🔥 Step: Drop rows where any of the important columns are null or blank
columns_to_clean = ['summary', 'timestamp', 'title', 'link', 'source']

# Convert whitespace-only strings to NA
for col in columns_to_clean:
    df[col] = df[col].astype(str).replace(r'^\s*$', pd.NA, regex=True)

# Drop rows where any of these columns are NA
df = df.dropna(subset=columns_to_clean, how='any')

# Reset index
df = df.reset_index(drop=True)

print(f"✅ Cleaned DataFrame: {df.shape[0]} rows remaining after removing null/blank in {columns_to_clean}")
```

# Convert timestamp to simplified datetime

```python
# Drop spam entries in timestamp
df = df[~df['timestamp'].str.lower().str.contains("thank you very much and god bless you", na=False)]
# Reset index and save
df = df.reset_index(drop=True)
```

```python
df["datetime"] = pd.to_datetime(df["timestamp"], errors="coerce", utc=True)
```

```python
df = df[df["datetime"].notna()]  # Remove rows with NaT

# Format the datetime as string: YYYY-MM-DD HH:MM
df["datetime"] = df["datetime"].dt.strftime("%Y-%m-%d %H:%M")
```

# Clean title and summary (remove "LIVE", trim spaces, etc.)

```python
# Step 1: Replace empty strings or whitespace-only strings with NaN
df['title'] = df['title'].replace(r'^\s*$', pd.NA, regex=True)
df['summary'] = df['summary'].replace(r'^\s*$', pd.NA, regex=True)

# Step 2: Drop rows where both title and summary are now NA
df = df.dropna(subset=['title', 'summary'], how='all')

# Step 3: Reset index
df = df.reset_index(drop=True)
```

```python
import re

def clean_text(text):
    if pd.isnull(text):
        return ""
    text = re.sub(r"\bLIVE\b", "", text, flags=re.IGNORECASE)  # remove "LIVE"
    text = re.sub(r"http\S+", "", text)                        # remove URLs
    text = re.sub(r"[^\w\s.,!?-]", "", text)                   # remove weird chars
    return text.strip()

df["title_clean"] = df["title"].apply(clean_text)
df["summary_clean"] = df["summary"].fillna("").apply(clean_text)
```

```python
# Count how many rows have "No summary available"
no_summary_count = df[df["summary_clean"] == "No summary available"].shape[0]
print(f"🕵️ Rows with 'No summary available': {no_summary_count}")
```

```python
# Drop rows where summary_clean is exactly "No summary available"
df = df[df["summary_clean"] != "No summary available"].reset_index(drop=True)
print(f"✅ DataFrame after removal: {df.shape[0]} rows remaining")
```

# Deduplication using MinHash

```python
from datasketch import MinHash, MinHashLSH

def get_minhash(text):
    m = MinHash(num_perm=128)
    for word in text.split():
        m.update(word.encode('utf8'))
    return m

lsh = MinHashLSH(threshold=0.85, num_perm=128)
unique_indices = []

for idx, row in df.iterrows():
    m = get_minhash(row["text"])
    if not any(lsh.query(m)):
        lsh.insert(f"text_{idx}", m)
        unique_indices.append(idx)

df = df.loc[unique_indices].reset_index(drop=True)

```

```python
# Step 2: Import spaCy and load the model
import spacy

# Load small English model
nlp = spacy.load("en_core_web_sm")

# Step 3: Define NER function to extract (entity, label) pairs
ALLOWED_ENTITY_TYPES = ["PERSON", "ORG", "GPE", "DATE", "TIME", "MONEY", "PERCENT", "NORP", "EVENT", "LAW", "FAC", "PRODUCT"]  # You can add more

def extract_filtered_entities(text):
    if not isinstance(text, str) or not text.strip():
        return []
    doc = nlp(text)
    return list(set((ent.text, ent.label_) for ent in doc.ents if ent.label_ in ALLOWED_ENTITY_TYPES))

# Step 4: Apply the function to the 'text' column
df["entities"] = df["text"].apply(extract_filtered_entities)

# Step 5: Preview result
df[["text", "entities"]].head()

```

# Separate Entity Text and Label

```python
df["entity_names"] = df["entities"].apply(lambda ents: [e[0] for e in ents])
df["entity_types"] = df["entities"].apply(lambda ents: [e[1] for e in ents])
```

# Keyphrase Extraction (TF-IDF)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Vectorize using unigrams + bigrams, remove common stop words
vectorizer = TfidfVectorizer(ngram_range=(1, 2), stop_words="english")
tfidf_matrix = vectorizer.fit_transform(df["text"])
feature_names = vectorizer.get_feature_names_out()

# Extract top N keywords per row
def get_top_keywords(row_index, top_n=5):
    row = tfidf_matrix[row_index].toarray().flatten()
    top_indices = row.argsort()[-top_n:][::-1]  # top 5 in descending order
    return [feature_names[i] for i in top_indices]

# Apply to all rows
df["keyphrases"] = [get_top_keywords(i) for i in range(len(df))]

```

# Language Detection

```python
from langdetect import detect

def detect_lang(text):
    try:
        return detect(text)
    except:
        return "unknown"

df["language"] = df["text"].apply(detect_lang)
```

```python
from transformers import MarianMTModel, MarianTokenizer
from tqdm import tqdm

# Enable progress bar in DataFrame apply
tqdm.pandas()

# Load model and tokenizer
model_name = "Helsinki-NLP/opus-mt-mul-en"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)
```

```python
def translate_text(text):
    try:
        tokens = tokenizer([text], return_tensors="pt", truncation=True, padding=True)
        translated = model.generate(**tokens)
        return tokenizer.decode(translated[0], skip_special_tokens=True)
    except Exception as e:
        print(f"Translation failed: {e}")
        return text  # Return original text on error
```

```python
df["translated_text"] = df.progress_apply(
    lambda x: translate_text(x["text"]) if x["language"] != "en" else x["text"],
    axis=1
)
```
