```python
import os
from PIL import Image

input_dir = '/content/drive/MyDrive/AIP_Project/week 5/final_resized_normalized'
output_dir = '/content/drive/MyDrive/AIP_Project/week 5/processed_emoji'

# Create the output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# Resize and preprocess images
for filename in os.listdir(input_dir):
    if filename.endswith(('png', 'jpg', 'jpeg')):
        img_path = os.path.join(input_dir, filename)
        img = Image.open(img_path).convert("RGB")  # Convert to RGB
        img = img.resize((64, 64))  # Resize to 64x64, or adjust to (128, 128) if needed
        img.save(os.path.join(output_dir, filename))

print("Preprocessing complete! Images saved to:", output_dir)
```

## Train StyleGAN2


```python
!ls "/content/drive/MyDrive/AIP_Project/week 5"
```

```python
from stylegan2_pytorch import Trainer

# Initialize StyleGAN2 Trainer
trainer = Trainer(
    name='emoji_stylegan',  # The name of the model for saving checkpoints
    results_dir='./results',  # Directory to store generated images and checkpoints
    image_size=64,  # Training image size
    batch_size=16,  # Adjust based on GPU memory
    num_workers=2,
    gradient_accumulate_every=4
)

# Set the data source to your processed emoji folder
trainer.set_data_src('/content/drive/MyDrive/AIP_Project/week 5/processed_emoji')
```

```python
steps = 1000  # Number of training steps
current_step = 0  # Initialize step counter

while current_step < steps:
    trainer.train()  # Train for a batch
    current_step += 1  # Increment the step count

    # Optional: Print progress every 100 steps
    if current_step % 100 == 0:
        print(f"Training step {current_step}/{steps} completed.")
```
